```python
def show_menu():
    print("\n--- TO-DO LIST ---")
    print("1. View Tasks")
    print("2. Add Task")
    print("3. Update Task")
    print("4. Delete Task")
    print("5. Mark as Done")
    print("6. Exit")

def view_tasks(tasks):
    if not tasks:
        print("\nNo tasks yet!")
        return
    print("\nYour Tasks:")
    for i, task in enumerate(tasks, 1):
        status = "✔ Done" if task["done"] else "❌ Pending"
        print(f"{i}. {task['name']} - {status}")

def add_task(tasks):
    name = input("Enter task name: ")
    tasks.append({"name": name, "done": False})
    print("Task added successfully!")

def update_task(tasks):
    view_tasks(tasks)
    if not tasks:
        return
    idx = int(input("Enter task number to update: ")) - 1
    if 0 <= idx < len(tasks):
        new_name = input("Enter new task name: ")
        tasks[idx]["name"] = new_name
        print("Task updated!")
    else:
        print("Invalid task number.")

def delete_task(tasks):
    view_tasks(tasks)
    if not tasks:
        return
    idx = int(input("Enter task number to delete: ")) - 1
    if 0 <= idx < len(tasks):
        removed = tasks.pop(idx)
        print(f"Deleted task: {removed['name']}")
    else:
        print("Invalid task number.")

def mark_done(tasks):
    view_tasks(tasks)
    if not tasks:
        return
    idx = int(input("Enter task number to mark done: ")) - 1
    if 0 <= idx < len(tasks):
        tasks[idx]["done"] = True
        print("Task marked as done!")
    else:
        print("Invalid task number.")

def main():
    tasks = []
    while True:
        show_menu()
        choice = input("Enter your choice: ")

        if choice == "1":
            view_tasks(tasks)
        elif choice == "2":
            add_task(tasks)
        elif choice == "3":
            update_task(tasks)
        elif choice == "4":
            delete_task(tasks)
        elif choice == "5":
            mark_done(tasks)
        elif choice == "6":
            print("Goodbye!")
            break
        else:
            print("Invalid choice, try again.")

if __name__ == "__main__":
    main()
```
